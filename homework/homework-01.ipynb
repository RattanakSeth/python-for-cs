### 4. Dataset description

The Titanic dataset is a CSV file that includes information about passengers aboard the Titanic. The goal is often to predict whether a passenger survived the sinking of the ship based on features such as:

+ PassengerId: A unique ID for each passenger.
+ Survived: 0 if the passenger did not survive, 1 if they did.
+ Pclass: The class of the passenger (1st, 2nd, or 3rd class).
+ Name: The passenger's name.
+ Sex: The gender of the passenger.
+ Age: The age of the passenger.
+ SibSp: Number of siblings/spouses aboard the Titanic.
+ Parch: Number of parents/children aboard the Titanic.
+ Ticket: The ticket number.
+ Fare: The amount paid for the ticket.
+ Cabin: The cabin number.
+ Embarked: The port where the passenger embarked (C = Cherbourg, Q = Queenstown, S = Southampton).<br>

This dataset is widely used for classification tasks, and it's a great resource for understanding basic concepts like data cleaning, feature engineering, and model evaluation.

### 5. Insight of titanic dataset

Here are some key insights and patterns often explored within this dataset:

1. Survival Rates
Overall Survival Rate: You can determine what percentage of passengers survived (e.g., ~38% survived).
Survival by Class: A key insight is that survival rates were much higher for passengers in higher classes. For example:
First class passengers had a significantly higher chance of survival compared to those in third class.
Survival by Gender: Women had a much higher survival rate compared to men, reflecting the "women and children first" policy.
2. Impact of Gender on Survival
Gender Analysis: Females were more likely to survive compared to males.
Survival rate for females: around 74%
Survival rate for males: around 19%
This provides insight into the importance of gender in survival predictions.
3. Impact of Age on Survival
Age Distribution: Children (younger passengers) had a higher chance of survival, particularly among females.
Age Binning: Binning ages into categories (e.g., children, adults, elderly) can further highlight survival trends based on age.
4. Impact of Class (Pclass)
Class and Wealth: Passengers in higher classes (Pclass = 1) had a higher survival rate.
First-class: ~63% survival rate.
Second-class: ~47% survival rate.
Third-class: ~24% survival rate.
Wealth and Safety: Higher class was an indicator of wealth, and wealthier passengers were more likely to be rescued.
5. Impact of Fare
Fare Amount: The fare paid by passengers was positively correlated with survival. Higher fare often meant a better class and, consequently, a better chance of survival.
The distribution of fare can also indicate socio-economic differences among the passengers.
6. Embarkation Point (Embarked)
Survival by Embarked Location: Passengers who embarked at different locations (Cherbourg, Queenstown, Southampton) showed different survival rates.
Passengers who embarked from Cherbourg (C) had a higher survival rate, while those from Southampton (S) had the lowest.
This may be related to the class distribution of passengers from each port.
7. Family Connections (SibSp & Parch)
Family Size Impact: Having more family members aboard could affect survival chances.
Passengers with smaller families (i.e., fewer siblings or parents onboard) tended to have higher survival rates.
Large families may have faced challenges sticking together or finding enough lifeboat seats.
8. Cabin Information
Cabin Data: Although many cabin entries are missing, cabin information can reveal patterns related to the passenger’s location on the ship (e.g., proximity to lifeboats).
9. Missing Data
The dataset has missing values, especially for fields like Age and Cabin. Dealing with missing data is part of data preprocessing and can offer insights into how to handle such issues in machine learning projects.
10. Machine Learning Insights
Feature Importance: By building models like logistic regression, decision trees, or random forests, you can assess the importance of features in predicting survival.
Commonly important features: Sex, Pclass, Fare, and Age.

```python
# Importing necessary libraries
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# After clean and analyse data
titanic_data = titanic #pd.read_csv(url)

# Preview the dataset
titanic_data.head()

# Calculating the overall survival rate
overall_survival_rate = titanic_data['survived'].mean() * 100
print(f"Overall Survival Rate: {overall_survival_rate:.2f}%")

# Survival rate by Gender
gender_survival_rate = titanic_data.groupby('sex')['survived'].mean() * 100
print("\nSurvival Rate by Gender:")
print(gender_survival_rate)

# Survival rate by Passenger Class (Pclass)
pclass_survival_rate = titanic_data.groupby('pclass')['survived'].mean() * 100
print("\nSurvival Rate by Passenger Class:")
print(pclass_survival_rate)

# Survival rate by Age Group (creating age bins)
age_bins = [0, 12, 18, 35, 60, 80]
age_labels = ['Child', 'Teenager', 'Adult', 'Middle-aged', 'Senior']
titanic_data['AgeGroup'] = pd.cut(titanic_data['age'], bins=age_bins, labels=age_labels)

agegroup_survival_rate = titanic_data.groupby('AgeGroup')['survived'].mean() * 100
print("\nSurvival Rate by Age Group:")
print(agegroup_survival_rate)

# Visualization of survival rates by Gender, Class, and Age Group
plt.figure(figsize=(12, 6))

# Plot 1: Survival Rate by Gender
plt.subplot(1, 3, 1)
sns.barplot(x=gender_survival_rate.index, y=gender_survival_rate.values)
plt.title('Survival Rate by Gender')
plt.ylabel('Survival Rate (%)')

# Plot 2: Survival Rate by Class
plt.subplot(1, 3, 2)
sns.barplot(x=pclass_survival_rate.index, y=pclass_survival_rate.values)
plt.title('Survival Rate by Class')
plt.ylabel('Survival Rate (%)')

# Plot 3: Survival Rate by Age Group
plt.subplot(1, 3, 3)
sns.barplot(x=agegroup_survival_rate.index, y=agegroup_survival_rate.values)
plt.title('Survival Rate by Age Group')
plt.ylabel('Survival Rate (%)')

plt.tight_layout()
plt.show()
```
