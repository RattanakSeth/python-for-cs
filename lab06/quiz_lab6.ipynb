```python
from matplotlib import pyplot as plt
%matplotlib inline
```

```python
"""
1. Line Graph
Data: Monthly Sales Performance of a Product
- Month: January, February, March, April, May
- Sales: 1000, 1500, 1200, 1800, 2000
"""

month = ["January", "February", "March", "April", "May"]
sales = [1000, 1500, 1200, 1800, 2000]
label = "Monthly Sales Performance of a Product"

plt.plot(month, sales)
plt.title(label=label)
plt.xlabel('Month')
plt.ylabel('Sales')
```

```python
"""
2. Bar Chart
Data: Annual Revenue Comparison of Multiple Products
- Product: A, B, C, D
- Revenue (in thousands): 250, 300, 400, 350
"""

label = 'Annual Revenue Comparison of Multiple Products'
product = ["A", "B", "C", "D"]
revenue = [250, 300, 400, 350]

plt.pie(revenue, labels=product)
plt.title(label=label)
plt.legend(title = "Products:")
# plt.xlabel('Product')
# plt.ylabel('Revenue')
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Apply style
plt.style.use('fivethirtyeight')

# Data
title = "Relationship between Study Hours and Test Scores"
hours = [2, 4, 3, 6, 5]
test_score = [60, 70, 65, 80, 75]
studyClass = pd.DataFrame({"hours": hours, "test_score": test_score})

# Plot
ax = studyClass.plot(x="hours", y="test_score", kind='scatter', title=title)
ax.set(xlabel="Study Hours", ylabel="Test Scores")

# Display the plot
plt.show()
```

```python
"""
4. Pie Chart
Data: Market Share of Different Mobile Brands
- Brand: Apple, Samsung, Xiaomi, Huawei, Others
- Market Share (in percentage): 40, 30, 15, 10, 5
"""

title = "Market Share of Different Mobile Brands"
brand = ["Apple", "Samsung", "Xiaomi", "Huawei", "Others"]
market = [40, 30, 15, 10, 5]

mkshare = pd.DataFrame({"brand": brand, "market": market})
ax = mkshare.plot(y="market", kind="pie", labels= mkshare['brand'], autopct = '%1.0f%%', legend=False, title='Market Share')

# Hide y-axis label
ax.set(ylabel='')
```

```python
"""
5. Area Chart
Data: Daily Website Traffic Over a Week
- Day: Monday, Tuesday, Wednesday, Thursday, Friday,
Saturday, Sunday
- Traffic (in thousands): 2, 3, 4, 3.5, 4.5, 5, 2.5
"""
# Data
title = "Daily Website Traffic Over a Week"
day = ["Monday", "Tuesday", "Wednesday", "Thursday", "Friday", "Saturday", "Sunday"]
traffic = [2, 3, 4, 3.5, 4.5, 5, 2.5]
webTraffic = pd.DataFrame({"day": day, "traffic": traffic})

# Plot
ax = webTraffic.plot(x="day", y="traffic", kind='area', title=title)

# Customize the plot
ax.set(xlabel="Day", ylabel="Traffic (in thousands)")

# Rotate x-axis labels to avoid overlap
plt.xticks(rotation=45)

# Show plot
plt.show()

```

```python
#06
"""
6. Histogram
Data: Distribution of Students' Exam Scores
-Score Range: 60-69, 70-79, 80-89, 90-99
-Number of Students: 5, 10, 15, 8
"""

# Data
title = "Distribution of Students' Exam Scores"
score_ranges = ["60-69", "70-79", "80-89", "90-99"]
num_students = [5, 10, 15, 8]
exam_scores = pd.DataFrame({"Score Range": score_ranges, "Number of Students": num_students})

# Plot
ax = exam_scores.plot(x="Score Range", y="Number of Students", kind='bar', title=title, legend=False)

# Customize the plot
```

```python
"""
7. Bubble Chart
Data: Population Comparison of Countries
-Country: USA, China, India, Brazil, Russia
-Population (in millions): 330, 1400, 1350, 210, 145
"""
title = "Population Comparison of Countries"
country = ["USA", "China", "India", "Brazil", "Russia"]
population = [330, 1400, 1350, 210, 145]

population_data = pd.DataFrame({"country": country, "population": population})

# Plot
plt.figure(figsize=(8, 6))
plt.scatter(population_data["country"], population_data["population"], 
            s=population_data["population"], alpha=0.5, color='blue')

# Customize the plot
plt.title(title)
plt.xlabel("Country")
plt.ylabel("Population (in millions)")

# Show plot
plt.show()
```

```python
"""
8. Box and Whisker Plot
Data: Monthly Temperature Range in a City
-Month: January, February, March, April, May
-Temperature Range (in degrees Celsius): 5-10, 8-15,
10-18,12-20, 15-25
"""

# Data
title = "Monthly Temperature Range in a City"
months = ["January", "February", "March", "April", "May"]
temperature_ranges = [[5, 10], [8, 15], [10, 18], [12, 20], [15, 25]]

# Create a DataFrame for temperature data
temperature_df = pd.DataFrame(temperature_ranges, columns=["Min Temp", "Max Temp"], index=months)

# Create box plot
plt.figure(figsize=(8, 6))
plt.boxplot(temperature_ranges, labels=months)

# Customize the plot
plt.title(title)
plt.ylabel("Temperature (°C)")

# Show plot
plt.show()
```
