```python
class Book:
    def __init__(self, title, author, pages):
        self.title = title
        self.author = author
        self.pages = pages
    
    # Method to display book details
    def display_info(self):
        #print(f"Title: {self.title}, Author: {self.author}, Pages: {self.pages}")
        print("Title:", self.title, ", Author: ", self.author, ", Pages: ", self.pages)

    def is_long_book(self):
        if self.pages > 300: 
            return True
        else: 
            return False
```

```python
book1 = Book("Richard III", "William Shakespeare", 350)
book1.display_info()
book1.is_long_book()

book2 = Book("The Great Gatsby", "F. Scott Fitzgerald", 180)
book2.display_info()
book2.is_long_book()
```
